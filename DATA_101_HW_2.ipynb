# **Part 1**

```python
import numpy as np
```

```python
students = ["Aurora", "Ariel", "Moana", "Tiana", "Cinderella", "Mulan", "Belle", "Jasmine", "Rapunzel", "Merida"]
assessments = ["Homework", "Project", "Exam1", "Exam2", "Final"]
grades = np.zeros((10, 5))
weights = np.array([0.15, 0.15, 0.20, 0.20, 0.30])

for i in range(10):
    name = input("Enter student name: ")
    students.append(name)

    print("Enter grades for", name)

    for j in range(5):
        grades[i][j] = float(input(assessments[j] + ": "))

final_grades = grades @ weights
letter_grades = []

for grade in final_grades:
    if grade >= 90:
        letter_grades.append("A")
    elif grade >= 80:
        letter_grades.append("B")
    elif grade >= 70:
        letter_grades.append("C")
    elif grade >= 60:
        letter_grades.append("D")
    else:
        letter_grades.append("F")

        print("\nStudent Grades")
print("-------------------------")

for i in range(10):
    print(students[i], ":", final_grades[i], letter_grades[i])
```
